## View 1: count the selection, not the world

Bar lengths count rows in this editorial sample. They do not measure repository quality, adoption, scientific validity or the size of either GitHub collection. Each collection contributes nine inspected repositories. A balanced sample is a design choice, not evidence that the ecosystems are equally large.

**Method connection:** compare the catalog's Table (§3), Bar chart (§6) and Semantic network (§23). Tables retain exact claims, bars compress an inventory, and network edges show selected relationships. The connection matrix below is an additional comparison technique. What important qualifications disappear when the source table becomes two bars?

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
atlas.groupby("collection").size().plot.bar(ax=axes[0], color=["#296d8b", "#d88735"], rot=0)
rx.finish(axes[0], "Curated inspected sample", "Collection", "Repositories (rows)")
has_file = ~atlas.rights_evidence.str.contains("no license file|No repository-wide", case=False)
pd.Series({"Scoped license file found": int(has_file.sum()), "No general file / assertion only": int((~has_file).sum())}).plot.barh(ax=axes[1], color="#527d68")
rx.finish(axes[1], "Review evidence, not an asset clearance", "Repositories (rows)", "")
fig.tight_layout()
plt.show()
```

## View 2: an explicit connection matrix

A filled square means *this course's editor connected this repository to this lesson*. Zero means no selected connection, rather than an absent capability. A repository may serve several questions. The two cholera repositories, for example, have different licensing evidence even though they concern the same historical event.

```python
labels = ["Atlas", "Climate", "Sea level", "Cholera", "Privacy", "STAC", "Quakes", "Coral", "Story/GIS"]
matrix = np.zeros((len(atlas), 9), dtype=int)
for row, values in enumerate(atlas.lessons):
    for value in values.split(","):
        matrix[row, int(value) - 1] = 1
fig, ax = plt.subplots(figsize=(11, 9))
ax.imshow(matrix, cmap="Blues", vmin=0, vmax=1, aspect="auto")
ax.set(xticks=range(9), xticklabels=labels, yticks=range(len(atlas)), yticklabels=atlas.repository,
       title="Selected teaching connections — pinned inspected sample")
ax.tick_params(axis="x", labelrotation=35)
fig.tight_layout()
plt.show()
display(pd.DataFrame({"lesson": labels, "connected_repositories": matrix.sum(axis=0)}))
```

## View 3: from matrix to bipartite graph

The same binary relationships can become edges. Here the layout has a fixed left-to-right reading order, avoiding a force layout that could imply meaningful distances. The graph is more immediate for paths; the matrix is better for precise comparison across many rows. Neither representation establishes a software dependency or an endorsement.

```python
fig, ax = plt.subplots(figsize=(12, 9))
for i in range(len(atlas)):
    for j in np.flatnonzero(matrix[i]):
        ax.plot([0, 1], [i, j * (len(atlas)-1)/8], color="#7196a5", alpha=.35, lw=1)
ax.scatter(np.zeros(len(atlas)), np.arange(len(atlas)), color="#296d8b", zorder=3)
ax.scatter(np.ones(9), np.linspace(0, len(atlas)-1, 9), color="#d88735", zorder=3)
for i, name in enumerate(atlas.repository):
    ax.text(-.025, i, name, ha="right", va="center", fontsize=8)
for j, name in enumerate(labels):
    ax.text(1.025, j*(len(atlas)-1)/8, name, va="center")
ax.set(xlim=(-.02, 1.02), title="Repository → lesson: editorial connections, no distance metric")
ax.invert_yaxis()
ax.axis("off")
fig.tight_layout()
plt.show()
```

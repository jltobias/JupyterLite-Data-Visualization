```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'category': ['A','B','C','D','E'],
    'value': [12, 19, 8, 15, 22]
})
df
```

```python
ax = df.plot.bar(x='category', y='value', legend=False, title='Example values by category')
ax.set_ylabel('Value')
plt.show()
```
